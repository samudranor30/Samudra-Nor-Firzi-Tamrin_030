```python
import matplotlib.pyplot as plt

# ===============================
# 1. INPUT DATA
# ===============================
print("=== SIMULASI GERAK LURUS BERUBAH BERATURAN (GLBB) ===")
a = float(input("Masukkan percepatan a (m/s^2): "))
v0 = float(input("Masukkan kecepatan awal v0 (m/s): "))
T  = int(input("Masukkan waktu total simulasi (detik): "))

# ===============================
# 2. VARIABEL PENYIMPAN DATA
# ===============================
x0 = 0
waktu = []
posisi = []

# ===============================
# 3. PERULANGAN PERHITUNGAN
# ===============================
for t in range(T+1):
    x = x0 + v0*t + 0.5*a*(t**2)  # rumus GLBB
    waktu.append(t)
    posisi.append(x)

    # IF CONDITION
    if x > 50:
        print(f"  Peringatan: benda melewati 50 meter pada t = {t} s")

# ===============================
# 4. OUTPUT TABEL
# ===============================
print("\n=== DATA POSISI ===")
print(f"{'t (s)':<8} {'x (m)':<10}")
for t, x in zip(waktu, posisi):
    print(f"{t:<8} {x:<10.2f}")

# ===============================
# 5. PLOTTING GRAFIK
# ===============================
plt.figure(figsize=(8,5))
plt.plot(waktu, posisi, marker='o')
plt.title("Grafik Posisi Terhadap Waktu (GLBB)")
plt.xlabel("Waktu (detik)")
plt.ylabel("Posisi (meter)")
plt.grid(True)
plt.show()
```
